```python
import pandas as pd
```

```python
df=pd.read_csv("employee_data.csv")
df.head()
```

```python
#identify missing values
df.describe()
```

```python

df.isnull().sum()
```

```python
import numpy as np
```

```python
#replace ag column with mean
mean_age=np.mean(df["Age"])
```

```python
df["Age"]=df["Age"].fillna(df["Age"].mean())
```

```python
df["Age"]
```

```python
#replace misisng salary using median
df["Salary"]=df["Salary"].fillna(df["Salary"].median())
df["Salary"]
```

```python
#replace department column using the mode
df["Department"]=df["Department"].fillna(df["Department"].mode()[0])
```

```python
#replace performance_score with mean
df["PerformanceScore"]=df["PerformanceScore"].fillna(df["PerformanceScore"].mean())
```

```python
df.isnull().sum()
```

```python
#replace joiningdate with mode
df["JoiningDate"] = df["JoiningDate"].fillna(df["JoiningDate"].mode()[0])
```

```python
df["JoiningDate"].dtype
```

```python
df["JoiningDate"] = pd.to_datetime(df["JoiningDate"])
df["JoiningDate"]
```

```python
#new col joining year
df["JoiningYear"] = df["JoiningDate"].dt.year
```

```python
df.head()
```

```python
df["JoiningMonth"] = df["JoiningDate"].dt.month
```

```python
df["JoiningDay"] = df["JoiningDate"].dt.day
```

```python
#no of employee joined before 2019
a=(df["JoiningYear"]<2019).sum()
a
```

```python
#split...
#contact col ->  email  and Phone

df[["Email","Phone"]]=df["Contact"].str.split("|",expand=True)
```

```python
df.head()
```

```python
df = df.drop(columns=[("Email", "Phone")])
```

```python
df["Email"] = df["Email"].str.strip()
df["Phone"] = df["Phone"].str.strip()
```

```python
#drop originl contact
#df.drop("Contact", axis=1, inplace=True)
```

```python
# create new column bonue - 10% of the salary
df["Bonus"]=df["Salary"]*0.1
```

```python
df.head(3)
```

```python
#new col -> total sal=sal+bonus
df["TotalSalary"]=df["Salary"]+df["Bonus"]
```
